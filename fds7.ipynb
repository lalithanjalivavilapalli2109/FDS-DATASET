```python
import pandas as pd
```

```python
# Student details
df1 = pd.DataFrame({
    'Student_ID': [1, 2, 3, 4],
    'Name': ['Sai', 'Rahul', 'Priya', 'Kiran'],
    'Department': ['CSE', 'ECE', 'CSE', 'IT']
})

print("Student Details:")
print(df1)
```

```python
# Student marks
df2 = pd.DataFrame({
    'Student_ID': [1, 2, 3, 5],
    'Marks': [85, 78, 92, 67],
    'Grade': ['A', 'B', 'A+', 'C']
})

print("Student Marks:")
print(df2)
```

```python
inner_merge = pd.merge(df1, df2, on='Student_ID', how='inner')

print(inner_merge)
```

```python
left_merge = pd.merge(df1, df2, on='Student_ID', how='left')

print(left_merge)
```

```python
right_merge = pd.merge(df1, df2, on='Student_ID', how='right')

print(right_merge)
```

```python
outer_merge = pd.merge(df1, df2, on='Student_ID', how='outer')

print(outer_merge)
```

```python
df3 = pd.DataFrame({
    'Student_ID': [6, 7],
    'Name': ['Arjun', 'Meena'],
    'Department': ['ECE', 'IT']
})

vertical_concat = pd.concat([df1, df3], axis=0, ignore_index=True)

print(vertical_concat)
```

```python
df4 = pd.DataFrame({
    'Attendance': [90, 85, 95, 80]
})

horizontal_concat = pd.concat([df1, df4], axis=1)

print(horizontal_concat)
```

```python
# Set Student_ID as the index
student_details = df1.set_index('Student_ID')
student_marks = df2.set_index('Student_ID')

joined_df = student_details.join(student_marks, how='inner')

print(joined_df)
```

```python
print("Missing values:")
print(outer_merge.isnull().sum())
```

```python
outer_merge['Marks'] = outer_merge['Marks'].fillna(0)

print(outer_merge)
```

```python
sorted_df = outer_merge.sort_values(by='Marks', ascending=False)

print(sorted_df)
```
