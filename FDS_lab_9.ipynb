```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.DataFrame({
    'Student_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Department': ['CSE', 'ECE', 'CSE', 'IT', 'ECE'],
    'CGPA': [8.5, 7.8, 9.1, 8.2, 7.5]
})

df1
```

```python
df2 = pd.DataFrame({
    'Student_ID': [101, 102, 103, 106, 107],
    'Marks': [85, 78, 92, 88, 75],
    'Grade': ['A', 'B', 'A+', 'A', 'B']
})

df2
```

```python
df_hierarchical = df1.set_index(['Department', 'Student_ID'])

df_hierarchical
```

```python
df_hierarchical.loc['CSE']
```

```python
df_a = pd.DataFrame({
    'Name': ['A', 'B'],
    'Marks': [80, 85]
})

df_b = pd.DataFrame({
    'Name': ['C', 'D'],
    'Marks': [90, 75]
})

combined_vertical = pd.concat([df_a, df_b], ignore_index=True)

combined_vertical
```

```python
df_c = pd.DataFrame({
    'Name': ['A', 'B', 'C']
})

df_d = pd.DataFrame({
    'Marks': [80, 85, 90]
})

combined_horizontal = pd.concat([df_c, df_d], axis=1)

combined_horizontal
```

```python
merged_df = pd.merge(
    df1,
    df2,
    on='Student_ID',
    how='inner'
)

merged_df
```

```python
left_merge = pd.merge(
    df1,
    df2,
    on='Student_ID',
    how='left'
)

left_merge
```

```python
outer_merge = pd.merge(
    df1,
    df2,
    on='Student_ID',
    how='outer'
)

outer_merge
```

```python
melted_df = pd.melt(
    df1,
    id_vars=['Student_ID', 'Name'],
    value_vars=['Department', 'CGPA'],
    var_name='Attribute',
    value_name='Value'
)

melted_df
```

```python
pivot_table = pd.pivot_table(
    df1,
    values='CGPA',
    index='Department',
    aggfunc='mean'
)

pivot_table
```

```python
pivot_table2 = pd.pivot_table(
    df1,
    values='CGPA',
    index='Department',
    columns='Name',
    aggfunc='mean'
)

pivot_table2
```

```python
stacked = df1.set_index(['Student_ID', 'Name'])[['CGPA']].stack()

stacked
```

```python
unstacked = stacked.unstack()

unstacked
```

```python
sorted_df = df1.sort_values(
    by='CGPA',
    ascending=False
)

sorted_df
```

```python
grouped_df = df1.groupby('Department')['CGPA'].mean()

grouped_df
```

```python
print("Original DataFrame:")
display(df1)

print("\nMerged DataFrame:")
display(merged_df)

print("\nHierarchical Index:")
display(df_hierarchical)

print("\nPivot Table:")
display(pivot_table)

print("\nReshaped DataFrame:")
display(melted_df)
```
