```python
# Import libraries
from sklearn.datasets import load_iris
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Iris dataset
iris = load_iris()

# Create DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Display first 5 rows
print(df.head())

# Correlation matrix
corr = df.corr()

print("\nCorrelation Matrix:")
print(corr)

# Plot heatmap
plt.figure(figsize=(6,5))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Iris Dataset")
plt.show()
```

```python
# Import libraries
from sklearn.datasets import load_iris
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Iris dataset
iris = load_iris()

# Create DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target

# 1. Histogram
plt.figure(figsize=(5,4))
plt.hist(df['sepal length (cm)'], bins=10, color='skyblue', edgecolor='black')
plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.show()

# 2. Scatter Plot
plt.figure(figsize=(5,4))
plt.scatter(df['sepal length (cm)'], df['petal length (cm)'], color='green')
plt.title("Scatter Plot")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.show()

# 3. Box Plot
plt.figure(figsize=(5,4))
sns.boxplot(data=df.iloc[:, :4])
plt.title("Box Plot of Iris Features")
plt.show()

# 4. Line Plot
plt.figure(figsize=(6,4))
plt.plot(df['sepal length (cm)'][:20], marker='o')
plt.title("Line Plot of Sepal Length")
plt.xlabel("Sample Index")
plt.ylabel("Sepal Length")
plt.show()

# 5. Bar Plot (Mean Values)
mean_values = df.iloc[:, :4].mean()
plt.figure(figsize=(6,4))
plt.bar(mean_values.index, mean_values.values, color='orange')
plt.title("Mean of Iris Features")
plt.xticks(rotation=20)
plt.ylabel("Mean")
plt.show()

# 6. Pair Plot
sns.pairplot(df, hue='species')
plt.show()
```
